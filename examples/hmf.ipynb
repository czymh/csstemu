# Halo mass function

Masses are in Msun/h; differential and cumulative number densities are in (Mpc/h)^-3. Redshifts and masses must be increasing.

```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

project_root = Path.cwd()
if not (project_root / 'CEmulator').is_dir():
    project_root = project_root.parent
sys.path.insert(0, str(project_root))
```

```python
from CEmulator.Emulator import HMF_CEmulator
emu = HMF_CEmulator(verbose=False)
emu.set_cosmos(Omegab=0.0486, Omegac=0.2604, H0=68.98,
               As=2.1e-9, ns=0.9665, mnu=0.06, checkbound=True)
z = np.array([0.0, 0.5, 1.0])
mass = np.logspace(11.5, 15, 50)
dndlnM = emu.get_dndlnM(z=z, M=mass, massdef='RockstarM200m')
number_density = emu.get_Nhalo(z=z, M=mass, massdef='RockstarM200m')
print(dndlnM.shape, number_density.shape)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for i, zi in enumerate(z):
    axes[0].loglog(mass, dndlnM[i], label=f'z={zi}')
    axes[1].loglog(mass, number_density[i], label=f'z={zi}')
for ax in axes:
    ax.set_xlabel('M [Msun/h]')
    ax.legend()
axes[0].set_ylabel('dn/dlnM [(Mpc/h)^-3]')
axes[1].set_ylabel('n(>M) [(Mpc/h)^-3]')
fig.tight_layout()
```
