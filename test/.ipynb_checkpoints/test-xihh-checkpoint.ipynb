```python
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
import matplotlib.pyplot as plt
import numpy as np
```

# Simple Usage

```python
import sys
sys.path.append('../')
from CEmulator.Emulator import Xihh_CEmulator
csstemu = Xihh_CEmulator(verbose=True)

rbins = np.concatenate([np.logspace(-2, 1, 30+1)[:-1],
                                np.arange(10, 500, 5)])
rmid  = (rbins[1:] + rbins[:-1]) / 2
rind  = (rmid <= 50 ) & (rmid >= 0.1 )

ic       = 0
zlists   = 1/np.loadtxt("/home/chenzhao/csst/simulation/zs_list12.txt") - 1
cosmoall = np.load("/home/chenzhao/csst/simulation/data/cosmologies_8d_train_n129_Sobol.npy")
Omegacb  = cosmoall[ic,1] - cosmoall[ic,7]/93.14/cosmoall[ic,2]/cosmoall[ic,2]*1e4
csstemu.set_cosmos(Omegab=cosmoall[ic,0], Omegac=cosmoall[ic,1]-cosmoall[ic,0], H0=cosmoall[ic,2],
                    ns=cosmoall[ic,3], As=cosmoall[ic,4]*1e-9, w=cosmoall[ic,5], 
                    wa=cosmoall[ic,6], mnu=cosmoall[ic,7])
xihh_tree = csstemu._get_xihh_tree(r=rmid, z=zlists[::-1], lgden1=-2.5, lgden2=-2.5, massdef="RockstarM200m")[::-1]
xihh_dire = csstemu.Xihh.get_xihh(r=rmid, z=zlists[::-1], lgden1=-2.5, lgden2=-2.5)[::-1]
xihh_comb = csstemu.get_xihh_lgnbar_threshold(r=rmid, z=zlists[::-1], lgden1=-2.5, lgden2=-2.5)[::-1] 
```

```python
plt.loglog(rmid,  xihh_comb[6], label='combine', color='C0')
plt.loglog(rmid, -xihh_comb[6], color='C0')
plt.loglog(rmid,  xihh_tree[6], ls='--', label='tree', color='C1')
plt.loglog(rmid, -xihh_tree[6], ls='--', color='C1')
plt.loglog(rmid,  xihh_dire[6], ls=':', label='direct', color='C2')
plt.loglog(rmid, -xihh_dire[6], ls=':', color='C2')
plt.legend()
plt.grid(True)
```

```python
xihh_all = np.load("/home/chenzhao/csst/simulation/analysis/halo-clustering/data/xihh_simu_n129_HybridNBin127_rockstar.npy")[0]
plt.loglog(rmid,  xihh_comb[6], label='Combine', color='C0')
plt.loglog(rmid, -xihh_comb[6], color='C0')
plt.loglog(rmid,  xihh_all [6,-1], ls=':', label='Simulation', color='C1')
plt.loglog(rmid, -xihh_all [6,-1], ls=':', color='C1')
plt.legend()
plt.grid(True)
```
